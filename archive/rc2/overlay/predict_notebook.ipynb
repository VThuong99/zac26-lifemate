# Frozen LifeMate K timing
Run all cells. Uses the same worker and transport as predict.sh. Time is milliseconds, starts before audio IO and includes ASR, all model decisions, ToolEnv calls and the final hash. Model loading is reported separately. Default output: /result/jupyter_submission.jsonl. No saved predictions are consumed.

```python
import os
from pathlib import Path
from predict import seed_everything, KRuntime, read_cases, write_predictions
seed_everything()
```

```python
data_root = Path(os.environ.get('LIFEMATE_DATA_ROOT', '/data'))
output = Path(os.environ.get('LIFEMATE_NOTEBOOK_OUTPUT', '/result/jupyter_submission.jsonl'))
runtime = KRuntime(data_root, os.environ.get('LIFEMATE_SPLIT'), output.parent / (output.stem + '.diagnostics'))
runtime.load()
```

```python
limit = int(os.environ['LIFEMATE_TEST_LIMIT']) if 'LIFEMATE_TEST_LIMIT' in os.environ else None
cases = read_cases(data_root, runtime.split, limit)
print('Requests:', len(cases))
```

```python
try:
    write_predictions(runtime, cases, output)
finally:
    runtime.close()
print('Wrote', output)
```
